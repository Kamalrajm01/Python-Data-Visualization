## Lists

#### Add an item to the list using append()

```python
Department_Store = ["Water Bottles", "Towels", "Grocery", "Diary Products", "Tiffin Box"]

Department_Store.append("Chocolates")
Department_Store
```

#### Add all elements of lst2 into lst1 using extend()

```python
Cars =["BMW","BENZ","AUDI","PORSCHE"]

Cars_2 =["TATA","HYUNDAI","KIA","MAHENDRA"]

Cars.extend(Cars_2)

Cars
```

#### Insert value at a specific position using insert()

```python
Cars.insert(5,"MARUTI SUZUKI")
```

```python
Cars
```

#### Remove a specific item using remove()

```python
Cars.remove("HYUNDAI")
```

```python
Cars
```

#### Remove last element using pop()

```python
Cars.pop(1)

Cars
```

#### Return index of value using index()

```python
Cars.index("BMW")
```

#### Count occurrences using count()

```python
Cars
Cars.count("TATA")
```
