```python
import pandas as pd
```

```python
Product_Data =pd.DataFrame({"Product ID" : [100,101,102,104,105], "Products" :["Mobiles","Laptop", "Mouse","AC","Printers"],"Quantity" : [5,4,8,7,6]})
```

```python
Product_Data
```

```python
Product_Price =pd.DataFrame({"Product ID" : [100,102,103,105,106], "Rate" :[80000,90000,500,45000,9500]})
```

```python
Product_Price
```

#### Inner 

```python
Sales = pd.merge(Product_Data,Product_Price,on= "Product ID",how="inner")
```

```python
Sales
```

#### left

```python
Sales = pd.merge(Product_Data,Product_Price,on= "Product ID",how="left")
```

```python
Sales
```

#### right

```python
Sales = pd.merge(Product_Data,Product_Price,on= "Product ID",how="right")
```

```python
Sales
```

#### outer

```python
Sales= pd.merge(Product_Data,Product_Price,on="Product ID",how="outer")
```

```python
Sales
```

#### Cross

```python
Sales= pd.merge(Product_Data,Product_Price,how="cross")
```

```python
Sales
```
