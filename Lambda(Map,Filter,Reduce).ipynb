#### Map

#### Example 1 Use map() to multiply all by 3

```python
numbers = [5,10,15,20,25]

result = list(map(lambda x:x*3, numbers))

print(result)
```

#### Example 2 Convert words to uppercase

```python
words =["laptop","mouse"]

result =list(map(lambda x: x.upper(), words))

print(result)
```

#### Filter

#### Example 1  Keep values greater than 30

```python
numbers = [15,30,45,60,75]

result =list(filter(lambda x: x>30, numbers))

print(result)
```

#### Example 2 Keep even numbers

```python
numbers = [15,30,45,60,75]

result =list(filter(lambda x: x%2==0, numbers))

print(result)
```

#### Reduce

#### Example 1 Find total sum

```python
numbers = [15,30]

result= reduce(lambda x,y: x+y,numbers)

print(result)
```

#### Example 2 Find multiplication of all numbers

```python
numbers = [5,3]

result= reduce(lambda x,y: x*y,numbers)

print(result)
```
