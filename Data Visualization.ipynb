```python
import pandas as pd
import seaborn as sns
import matplotlib. pyplot as plt
```

```python
Tips_data = sns.load_dataset("tips")
Tips_data
```

```python
Tips_data.isna().sum()
```

```python
Tips_data.describe()
```

## Univariate Analysis

####  Histogram

```python
plt.hist(Tips_data.total_bill)
plt.xlabel("total_bill")
plt.ylabel("Frequency")
plt.title("Total_Bill")
plt.show()
```

#### Countplot

```python
sns.countplot(data=Tips_data, x="time")
plt.title("Lunch VS Dinner")
plt.show()
```

#### Pie Chart

```python
Tips_data["day"].value_counts()
```

```python
plt.pie(x =Tips_data["day"].value_counts(),labels =["Saturday","Sunday","Thursday","Friday"],colors = ["green", "pink", "brown", "red"],autopct="%1.1f%%")
plt.show()
```
