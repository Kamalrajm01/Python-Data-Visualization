#### Example 1

```python
Companies = ["walmart","microsoft","oracle","apple","amazon"]

upper_companies = []

for x in Companies:
    if x:
        upper_companies.append(x.upper())
        
print(upper_companies)
```

#### Example 2

```python
numbers = [10, 15, 22, 33, 40, 55, 60]

even_numbers =[]

for x in numbers:
    if x%2==0:
        even_numbers.append(x)
print(even_numbers)        
        
```
