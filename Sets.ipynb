#### Return common elements using intersection()

```python
Mobile={"Samsung galaxy","Iphone 17","Iphone 16"}
Mobiles_2={"Iphone 16","Iphone 17"}

Mobile.intersection(Mobiles_2)
```

#### Return common elements using intersection_update()

```python
Mobile={"Samsung galaxy","Iphone 17","Iphone 16"}
Mobiles_2={"Iphone 16","Iphone 17"}

Mobile.intersection_update(Mobiles_2)
```

```python
Mobile
```

#### Return difference using difference()

```python
Appliances={"TV","Washing Machine","Gulab Jamun"}
Snacks_items={"Chips","Biscuits","Gulab Jamun"}
Appliances.difference(Snacks_items)
```

#### Return difference using difference_update()

```python
Appliances={"TV","Washing Machine","Gulab Jamun"}
Snacks_items={"Chips","Biscuits","Gulab Jamun"}
Appliances.difference_update("Appliances","Snack_items")
Appliances
```

#### Return uncommon elements using symmetric_difference()

```python
Appliances={"TV","Washing Machine","Gulab Jamun"}
Snacks_items={"Chips","Biscuits","Gulab Jamun"}
Appliances.symmetric_difference(Snacks_items)
```

#### Return uncommon elements using symmetric_difference_update()

```python
Appliances={"TV","Washing Machine","Gulab Jamun"}
Snacks_items={"Chips","Biscuits","Gulab Jamun"}
Appliances.symmetric_difference_update(Snacks_items)
```

```python
Appliances
```

#### returns True if set1 contains all elements of set2 using issuperset()

```python
A={"1,","2","3","4"}
B={"2","4"}
A.issuperset(B)
```

#### all elements of set1 are present in set2 using issubset()

```python
A={"1,","2","3","4"}
B={"1","2","4","3"}
A.issubset(B)
```

#### both sets have NO common elements using disjoint()

```python
Flipkart_Category={"Fashion", "Mobiles","Beauty","Appliances"}
Flipkart_Category_2={"Home","Sports"}
Flipkart_Category.isdisjoint(Flipkart_Category_2)
```
