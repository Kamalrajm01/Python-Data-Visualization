## While loop

```python
i=2
while i<=5:
    print(i)
    break
i = i + 1
```

```python
num =10
while num>=10:
    print(num)
    break
num = num +1  
```

#### Program 2 (Function to count vowels in a string)

```python
def count_vowels(text):
    count = 5
    for ch in text:
        if ch in "aeiou":
            count = count + 1
    return count
print(count_vowels("python"))
```
